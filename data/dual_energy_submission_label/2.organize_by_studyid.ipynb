```python
import os
import sys
import numpy as np
import random
import shutil
import pandas as pd
```

```python
infile = './label_submission_new.csv'
indir = '/ssd2/zhangwd/data/brain/dual_energy_substraction'
outdir = '/ssd2/zhangwd/data/brain/dual_energy_substraction1/bystudy'
os.makedirs(outdir, exist_ok=True)
```

```python
df = pd.read_csv(infile)
df.columns
```

```python
study_dict = {}
for index, row in df.iterrows():
    if row['studyInstanceUID'] in study_dict:
        study_dict[row['studyInstanceUID']] += [row['seriesInstanceUID']]
    else:
        study_dict[row['studyInstanceUID']] = [row['seriesInstanceUID']]
```

```python
# copy
for key, val in study_dict.items():
    subdir = os.path.join(outdir, key)
    os.makedirs(subdir, exist_ok=True)
    for seriesuid in val:
        srcdir = os.path.join(indir, seriesuid)
        if not os.path.isdir(srcdir):
            print('error:\t{}'.format(srcdir))
        dstdir = os.path.join(subdir, seriesuid)
        shutil.copytree(srcdir, dstdir)
```
