## 统计数据的最大和最小层数，并根据相应需求，提前做好padding，保证数据预处理步骤更快
1. 统计原始dicom数据的层数
2. 统计做过等比例缩放后数据的层数

```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
from collections import  Counter
import SimpleITK as sitk
import nibabel as nib
import time
```

```python
def count_layers(series_uid):
    reader = sitk.ImageSeriesReader()
    filenames = reader.GetGDCMSeriesFileNames(series_uid)
    reader.SetFileNames(filenames)
    im = reader.Execute()
    return im.GetSize()[2]
```

```python
data_in_root = '../data/Liver'
layer_counts = []
for sub_root_name in os.listdir(data_in_root):
    sub_root = os.path.join(data_in_root, sub_root_name)
    if not os.path.isdir(sub_root):
        continue
    if '3Dircadb' not in sub_root:
        continue
    volume_in_series = os.path.join(sub_root, 'PATIENT_DICOM/PATIENT_DICOM')
    if not os.path.isdir(volume_in_series):
        print('volume not exist:\t{}'.format(volume_in_series))
        continue
    layers_count = count_layers(volume_in_series)
    layer_counts.append(layers_count)
    print('{} layers count:\t{}'.format(volume_in_series, layers_count))
```

```python
print('raw data max layers:\t{}'.format(max(layer_counts)))
print('raw data min layers:\t{}'.format(min(layer_counts)))
```

```python
data_in_root = '../data/processed_liver'
layer_counts = []
for sub_root_name in os.listdir(data_in_root):
    sub_root = os.path.join(data_in_root, sub_root_name)
    if not os.path.isdir(sub_root):
        continue
    if '3Dircadb' not in sub_root:
        continue
    volume_in_series = os.path.join(sub_root, 'scalex1/volume')
    if not os.path.isdir(volume_in_series):
        print('volume not exist:\t{}'.format(volume_in_series))
        continue
    series_uid = os.path.join(volume_in_series, 'volume.nii')
    image = sitk.ReadImage(series_uid)
    layers_count = image.GetSize()[2]
    layer_counts.append(layers_count)
    print('{} layers count:\t{}'.format(volume_in_series, layers_count))
    
print('raw data max layers:\t{}'.format(max(layer_counts)))
print('raw data min layers:\t{}'.format(min(layer_counts)))
```
