解压后，每个目录下仍有很多未解压的文件，再继续进行解压，如下：

````python
data_root = '../data/Liver'
for sub_root in os.listdir(data_root):
    sub_root = os.path.join(data_root, sub_root)
    print(sub_root)
    if not os.path.isdir(sub_root):
        continue
    if '3Dircadb' not in sub_root:
        continue
    to_unziped_files = glob(os.path.join(sub_root, '*.zip'))
    os.system('cd {}'.format(sub_root))
    for f in to_unziped_files:
        unzip_out_dir = f.replace('.zip', '')
        unzip_cmd = 'unzip {} -d {}'.format(f, unzip_out_dir)
        print(unzip_cmd)
        os.system(unzip_cmd)
````
