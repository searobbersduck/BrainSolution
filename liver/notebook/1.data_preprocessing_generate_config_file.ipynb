```python
import os
import sys
from glob import glob
```

```python
data_root = '../data/processed_liver_liver/scalex1_liver'

volume_inpattern = '_volume_[0-1].npy'
mask_inpattern = '_mask.npy'

volume_list = glob(os.path.join(data_root, '*_volume_[0-1*.npy'.format(volume_inpattern)))
print('volume list size:\t{}'.format(len(volume_list)))
mask_list = []
for volume_file in volume_list:
    mask_file = volume_file.replace(volume_inpattern, mask_inpattern)
    if not os.path.isfile(mask_file):
        continue
    mask_list.append(mask_file)

outfile = '../data/processed_liver_liver/liver_liver_seg_0_1.txt'
with open(outfile, 'w') as f:
    for i in range(len(volume_list)):
        f.write('{}\t{}\n'.format(os.path.basename(volume_list[i]), os.path.basename(mask_list[i])))
```
