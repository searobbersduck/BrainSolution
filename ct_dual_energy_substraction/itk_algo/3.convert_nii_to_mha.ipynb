```python
import SimpleITK as sitk
from glob import glob
import os
import sys
```

```python
def niigz_to_mha(root_dir):
    root_path = root_dir
    nii_path = root_path + '/mask_demo0.nii.gz'
    mha_path = root_path + '/mask_demo0.mha'
    img = sitk.ReadImage(nii_path)
    sitk.WriteImage(img, mha_path)
    print('convert from {} to {}'.format(nii_path, mha_path))
```

```python
root_dir = '/data/brain'
for sub_dir in os.listdir(root_dir):
    if len(sub_dir) < 10:
        continue
    sub_dir = os.path.join(root_dir, sub_dir)
    if not os.path.isdir(sub_dir):
        continue
    niigz_to_mha(sub_dir)
```
