```python
import os
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
%matplotlib inline
import SimpleITK
import itk
import json
import SimpleITK as sitk
import time

from ipywidgets import interact, interactive
from ipywidgets import widgets
```

```python
# 1. load volume data
series_uid = '/data/zhangwd/data/examples/brain/bystudy/1.3.12.2.1107.5.1.4.60320.30000016072900171834200000026/1.3.12.2.1107.5.99.2.9594.30000016072913081812500000799'
reader = sitk.ImageSeriesReader()
filenamesDICOM = reader.GetGDCMSeriesFileNames(series_uid)
print(len(filenamesDICOM))
reader.SetFileNames(filenamesDICOM)
imgOriginal = reader.Execute()

# 2. initial mask
seg = sitk.Image(imgOriginal.GetSize(), sitk.sitkUInt8)
seg.CopyInformation(imgOriginal)

# 3. set seed point, change mask
seed = (228, 195, 154)
seg[seed] = 1

# 4. dilate mask seed point
seg = sitk.BinaryDilate(seg, 2)
```
