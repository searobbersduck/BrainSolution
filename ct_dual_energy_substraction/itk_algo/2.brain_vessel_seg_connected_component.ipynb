```python
import os
import sys
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
%matplotlib inline
import SimpleITK as sitk
import time
```

```python
print(sitk.Version())
```

```python
series_uids = []
series_list_file = '/data/zhangwd/data/examples/brain/bone_removed/removed_dicom.txt'
with open(series_list_file) as f:
    for line in f.readlines():
        line = line.strip()
        if line is None or len(line) == 0:
            continue
        series_uids.append(line)
```

```python
series_uid = series_uids[0]
print(series_uid)
```

```python
reader = sitk.ImageSeriesReader()
filenamesDICOM = reader.GetGDCMSeriesFileNames(series_uid)
reader.SetFileNames(filenamesDICOM)
imgOriginal = reader.Execute()
```

```python
# image2DThresh = sitk.Threshold(imgOriginal, lower=100, upper=500)
# cca = sitk.ConnectedComponentImageFilter()
# cca_image = cca.Execute(image2DThresh)


def save_raw_connected_result(lower, upper, imgOriginal, display=False):
    image2DThresh = sitk.Threshold(imgOriginal, lower=lower, upper=upper)
    cca = sitk.ConnectedComponentImageFilter()
    cca_image = cca.Execute(image2DThresh)
    sitk.WriteImage(cca_image, './connected/connected_component_{}_{}.nii.gz'.format(lower, upper))
    
def save_raw_threshold_result(lower, upper, imgOriginal, display=False):
    image2DThresh = sitk.Threshold(imgOriginal, lower=lower, upper=upper)
#     cca = sitk.ConnectedComponentImageFilter()
#     cca_image = cca.Execute(image2DThresh)
    os.makedirs('./threshold', exist_ok=True)
    sitk.WriteImage(image2DThresh, './threshold/connected_component_{}_{}.nii.gz'.format(lower, upper))
    
os.makedirs('./connected',exist_ok=True)

lowers = list(range(100,500,20))
uppers = list(range(400,600,20))

filter = sitk.MinimumMaximumImageFilter()
filter.Execute(imgOriginal)
ImageHighestIntensity = filter.GetMaximum()
ImageLowestIntensity = filter.GetMinimum()
if display:
    print(ImageLowestIntensity)
    print(ImageHighestIntensity)

for lower in lowers:
    for upper in uppers:
        print('===> processing threshold[{}, {}]'.format(lower, upper))
        beg = time.time()
#         save_raw_connected_result(lower,upper, imgOriginal)
#         save_raw_threshold_result(lower,upper, imgOriginal)
        end = time.time()
        print('time elapsed:{:.3f}s'.format(end-beg))
```

```python
def save_connected_component_result(lower, upper, mImg, index=1, display=False):
    image2DThresh = sitk.Threshold(mImg, lower=lower, upper=upper)
    image2DThresh = sitk.Cast(image2DThresh, sitk.sitkInt16)
    sitk.WriteImage(image2DThresh, './connected1/connected_threshold_{}_{}.nii.gz'.format(lower, upper))
    cca = sitk.ConnectedComponentImageFilter()
    cca_image = cca.Execute(image2DThresh)
    sitk.WriteImage(cca_image, './connected1/connected_component_{}_{}.nii.gz'.format(lower, upper))

lowers = list(range(70,150,10))
uppers = list(range(170,300,20))

mImg = sitk.ReadImage('/data/ctCardiac1/1.3.12.2.1107.5.99.2.9594.30000017032111384964000000240/medialness.nii.gz')
filter = sitk.MinimumMaximumImageFilter()
filter.Execute(mImg)
ImageHighestIntensity = filter.GetMaximum()
ImageLowestIntensity = filter.GetMinimum()
if display:
    print(ImageLowestIntensity)
    print(ImageHighestIntensity)

os.makedirs('./connected1',exist_ok=True)
for lower in lowers:
    for upper in uppers:
        print('===> processing threshold[{}, {}]'.format(lower, upper))
        beg = time.time()
        save_connected_component_result(lower,upper, mImg)
        end = time.time()
        print('time elapsed:{:.3f}s'.format(end-beg))
```
