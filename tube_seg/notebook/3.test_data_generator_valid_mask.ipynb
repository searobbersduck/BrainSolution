```python
import os
import sys
sys.path.append('../')
from datasets.neuro_vascular_segment_dataset_valid_mask import NeuroVascularSegmentDS
import matplotlib.pyplot as plt
import numpy as np
from torch.utils.data import DataLoader
import SimpleITK as sitk
import shutil
```

```python
root_dir = '../data/brain_henan/algo_mask/brain_seg_by_threshold_preprocessed'
img_list = '../data/brain_henan/algo_mask/config/neuro_vascular_seg_mix_include_range.txt'
phase = 'train'
crop_size = [128,128,128]
scale_size = [128,128,128]

ds = NeuroVascularSegmentDS(root_dir, img_list, phase, crop_size, scale_size)

dataloader = DataLoader(ds, batch_size=1, num_workers=8, shuffle=True, pin_memory=False)

cnt_total = 0
cnt_valid = 0
outdir = '../data/brain_henan/tmp'
os.makedirs(outdir, exist_ok=True)
epoch = 0
for index, (images, masks, image_paths, _) in enumerate(dataloader):
    print('====> save to:\t{}'.format(index))
    writer = sitk.ImageFileWriter()
    writer.SetFileName(os.path.join(outdir, 'epoch_{}_index_{}_volume.nii.gz'.format(epoch, index)))
    writer.Execute(sitk.GetImageFromArray(images[0][0].numpy()))
    
    writer = sitk.ImageFileWriter()
    writer.SetFileName(os.path.join(outdir, 'epoch_{}_index_{}_mask.nii.gz'.format(epoch, index)))
    writer.Execute(sitk.GetImageFromArray(masks[0].numpy()))
    
    image_path = image_paths[0]
    image_path = image_path.replace('.npy', '.nii.gz')
    basename = os.path.basename(image_path)
    shutil.copyfile(image_path, os.path.join(outdir, 'epoch_{}_index_{}_{}'.format(epoch, index, basename)))
```
