```python
volume_inpattern = '_mix.npy'
mask_inpattern = '_mask.npy'

volume_list = glob(os.path.join(data_root, '*_mix.npy'.format(volume_inpattern)))
print('volume list size:\t{}'.format(len(volume_list)))
mask_list = []
# print(volume_list)
for volume_file in volume_list:
    mask_file = volume_file.replace(volume_inpattern, mask_inpattern)
    if not os.path.isfile(mask_file):
        continue
    mask_list.append(mask_file)

ranges = []
import numpy as np
for mask_file in mask_list:
    print('processing {}'.format(mask_file))
    mask = np.load(mask_file)
    zero_value = 0
    non_zeros_idx = np.where(mask != zero_value)
    [max_z, max_h, max_w] = np.max(np.array(non_zeros_idx), axis=1)
    [min_z, min_h, min_w] = np.min(np.array(non_zeros_idx), axis=1)
    print('{}\t{}\t{}\t{}\t{}\t{}'.format(min_z, max_z, min_h, max_h, min_w, max_w))
    ranges.append('{}\t{}\t{}\t{}\t{}\t{}'.format(min_z, max_z, min_h, max_h, min_w, max_w))
    
outfile = '../data/brain_henan/algo_mask/config/neuro_vascular_seg_mix_include_range.txt'
with open(outfile, 'w') as f:
    for i in range(len(volume_list)):
        f.write('{}\t{}\t{}\n'.format(os.path.basename(volume_list[i]), os.path.basename(mask_list[i]), ranges[i]))
```
