```python
import os
import sys
import SimpleITK as sitk
```

```python
img1 = sitk.ReadImage('../data/TestMeshToImage/LC-LA4.mhd')
img2 = sitk.ReadImage('../data/TestMeshToImage/LC-M3.mhd')
img3 = sitk.ReadImage('../data/TestMeshToImage/LC-P3.mhd')
img4 = sitk.ReadImage('../data/TestMeshToImage/RC-M3.mhd')
img5 = sitk.ReadImage('../data/TestMeshToImage/RC-RA4.mhd')
img6 = sitk.ReadImage('../data/TestMeshToImage/RV-P3.mhd')
nary_add_filter = sitk.NaryAddImageFilter()
```

```python
result_img = nary_add_filter.Execute(img1, img2, img3, img4)
result_img = nary_add_filter.Execute(result_img, img5, img6)

thres_filter = sitk.BinaryThresholdImageFilter()
thres_filter.SetInsideValue(1)
thres_filter.SetOutsideValue(0)
thres_filter.SetLowerThreshold(1)
thres_filter.SetUpperThreshold(100)
result_img = thres_filter.Execute(result_img)

writer = sitk.ImageFileWriter()
writer.SetFileName(os.path.join('../data/TestMeshToImage', 'merge.mha'))
writer.Execute(result_img)
```
