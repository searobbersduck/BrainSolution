```python
# 处理剪影数据
nii_root = '../data/brain_henan/nii/bone_removed'
os.makedirs(nii_root, exist_ok=True)
mhd_list = glob(os.path.join('../data/brain_henan/mhd/out', '*.mhd'))
for mhd_path in mhd_list:
    try:
        nii_path = os.path.join(nii_root, os.path.basename(mhd_path).replace('.mhd', '.nii'))
        img = sitk.ReadImage(mhd_path)
        sitk.WriteImage(img, nii_path)
    except:
        continue
```
