```python
import os
import sys
import SimpleITK as sitk

import numpy as np
```

```python
fixed_file = '../data/gan/ncct2dwi/experiment_registration2/2.nii_file_ori/469298_first_BS_brain.nii.gz'
moving_file = '../data/gan/ncct2dwi/experiment_registration2/2.nii_file_ori/469298_first_FU_DWI_BXXX.nii.gz'
```

```python
fixed_image = sitk.ReadImage(fixed_file, sitk.sitkFloat32)
moving_image = sitk.ReadImage(moving_file, sitk.sitkFloat32)


initial_transform = sitk.CenteredTransformInitializer(fixed_image, 
                                                      moving_image, 
                                                      sitk.Euler3DTransform(), 
                                                      sitk.CenteredTransformInitializerFilter.GEOMETRY)
```
